```python
# Corrigo los tipos de datos , convertir cadenas a fechas.
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df.dtypes
```

```python
# Crear nuevas columnas extrayendo información de la columna 'mes_venta'
df['Month'] = df['Date'].dt.month

print("Nueva columna creada:")
df['Month'].head(5)
```

```python
# Cambiar el numero por mes del año
month_map = {
    1: 'January', 2: 'February', 3: 'March', 4: 'April',
    5: 'May', 6: 'June', 7: 'July', 8: 'August',
    9: 'September', 10: 'October', 11: 'November', 12: 'December'
}
df['Month'] = df['Month'].map(month_map)

print("Columna 'Month' convertida a nombres de mes:")
print(df['Month'].head())
```

```python
# Clasifica los datos en categorías relevantes.
# Encontrando el índice del producto con mayor venta por categoría
idx = df.groupby('Product Category')['Total Amount'].idxmax()

# 'groupby' agrupa el DataFrame por la columna 'Product Category'
# 'idxmax' devuelve el índice de la fila con el valor máximo de ventas dentro de cada grupo (categoría)
# Esto nos da los índices del producto con mayor venta en cada categoría
# Luego, usamos 'loc' para seleccionar las filas correspondientes a esos índices
top_productos = df.loc[idx, ['Product Category', 'Gender', 'Total Amount']]

# Imprimiendo los resultados
print("Genero que mas compra una Categoría:")
print(top_productos)
```

### Clasifica los datos: Crea una columna que clasifique las ventas en categorías significativas (e.g., ‘Alta’, ‘Media’, ‘Baja’).

```python
# Creamos la nueva columna dividiendo en 3 grupos iguales (terciles)
df['Sales Category'] = pd.qcut(df['Total Amount'], q=3, labels=['Low', 'Med', 'Hig'])
```

### Crea gráficos de líneas para mostrar las tendencias de ventas a lo largo del tiempo.



```python
df.groupby(['Date'])['Total Amount'].sum()
```

```python
# Crear el gráfico de líneas
Sales_by_date = df.groupby(['Date'])['Total Amount'].sum()
plt.plot(Sales_by_date.index, Sales_by_date.values)
plt.title("Gráfico de Líneas de Ventas a lo largo del Tiempo")
plt.xlabel("Fecha")
plt.ylabel("Total Amount")
plt.grid(True)
plt.show()
```

```python
# Crear el gráfico de líneas con más variables
# Segmento las ventas por categoria para mostrar graficamente cual las ventas de cada una

# Calculo ventas de 'Electronics'
Electronics_sales_by_date = df[df['Product Category'] == 'Electronics'].groupby('Date')['Total Amount'].cumsum()

# Calculo ventas de 'Clothing'
Clothing_sales_by_date = df[df['Product Category'] == 'Clothing'].groupby('Date')['Total Amount'].cumsum()

# Calculo ventas de 'Beauty'
Beauty_sales_by_date = df[df['Product Category'] == 'Beauty'].groupby('Date')['Total Amount'].cumsum()

plt.plot(Electronics_sales_by_date.index, Electronics_sales_by_date.values, label='Electronics Sales')
plt.plot(Clothing_sales_by_date.index, Clothing_sales_by_date.values, label='Clothing Sales')
plt.plot(Beauty_sales_by_date.index, Beauty_sales_by_date.values, label='Beauty Sales')

plt.title("Gráfico de Líneas de Ventas por Categoría a lo largo del Tiempo")
plt.xlabel("Fecha")
plt.ylabel("Monto")
plt.legend()
plt.grid(True)
plt.show()
```

```python
fig, axs = plt.subplots(3, 1, figsize=(12, 10))


# Acumulado en ventas de 'Electronics'
Electronics_sales_by_date = df.sort_values(by="Date").reset_index(drop=True)
Electronics_sales_by_date = df[df['Product Category'] == 'Electronics'].groupby('Date')['Total Amount'].sum().cumsum()


# Acumulado en ventas de 'Clothing'
Clothing_sales_by_date = df.sort_values(by="Date").reset_index(drop=True)
Clothing_sales_by_date = df[df['Product Category'] == 'Clothing'].groupby('Date')['Total Amount'].sum().cumsum()

# Acumulado en ventas de 'Beauty'
Beauty_sales_by_date = df.sort_values(by="Date").reset_index(drop=True)
Beauty_sales_by_date = df[df['Product Category'] == 'Beauty'].groupby('Date')['Total Amount'].sum().cumsum()

# Subplot 1
axs[0].plot(Electronics_sales_by_date.index, Electronics_sales_by_date.values, 'tab:blue', label='Electronics Sales')
axs[0].set_title('Ventas de Electronics')

# Subplot 2
axs[1].plot(Clothing_sales_by_date.index, Clothing_sales_by_date.values, 'tab:orange', label='Clothing Sales')
axs[1].set_title('Ventas de Clothing')

# Subplot 3
axs[2].plot(Beauty_sales_by_date.index, Beauty_sales_by_date.values, 'tab:green', label='Beauty Sales')
axs[2].set_title('Ventas de Beauty')

# Ajustar la posición de los subplots para evitar superposición
plt.tight_layout()
plt.show()
```
